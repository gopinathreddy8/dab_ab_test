# Default notebook

This default notebook is executed using a Lakeflow job as defined in resources/sample_job.job.yml.

```python
# Set default catalog and schema
catalog = dbutils.widgets.get("catalog")
schema = dbutils.widgets.get("schema")
spark.sql(f"USE CATALOG {catalog}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {schema}")
spark.sql(f"USE SCHEMA {schema}")
```

```python
spark.sql(f"""CREATE or REPLACE TABLE {catalog}.{schema}.transactions_summary AS
SELECT  product,
paymentMethod,
sum(quantity) as total_quantity,
SUM(totalPrice) AS total_price
FROM `samples`.`bakehouse`.`sales_transactions`
group by product,paymentMethod""")
```
